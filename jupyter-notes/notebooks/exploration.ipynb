```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
rng = np.random.default_rng(42)
days = np.arange(1, 31)
values = 20 + 0.35 * days + rng.normal(0, 1.5, size=days.size)
data = pd.DataFrame({"day": days, "value": values})
data.head()
```

```python
fig, ax = plt.subplots(figsize=(9, 4.5))
sns.lineplot(data=data, x="day", y="value", marker="o", ax=ax)
ax.set(title="Example trend", xlabel="Day", ylabel="Value")
plt.tight_layout()
```

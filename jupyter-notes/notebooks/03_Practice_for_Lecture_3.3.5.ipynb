### NEES calculation



```python
import numpy as np

y_tilde = np.array([2.0, -1.0]) # Innovation
#print(y_tilde)

Sigma_y = np.array([[2.0, 1.0], [1.0, 2.0]]) # Innovation covariance
#print(Sigma_y)

nees = y_tilde.T @ np.linalg.inv(Sigma_y) @ y_tilde
print("NEES:", nees)
```

## Oberen X^2 Grenzwert berechnen

mit Restwahrscheinlichkeit alpha = 2% = 0.02 und m = 1 Freiheitsgraden




```python
from scipy.stats import chi2

alpha = 0.025
m = 2

chi_upper = chi2.ppf(1 - alpha, df =  m)

print("Chi-squared upper critical value:", chi_upper)
```
